```python
df = df.drop_duplicates()
print("Duplicate Rows Removed")
```

```python
for col in df.columns:
    if df[col].isnull().any():
        if pd.api.types.is_numeric_dtype(df[col]):
            df[col] = df[col].fillna(df[col].mean())
        else:
            df[col] = df[col].fillna(df[col].mode()[0])

print("\nMissing Values Handled")
```

```python
#1.
print(df[(df["Sex"] == "female") & (df["Survived"] == 1) & (df["Age"] > 30)][["Name", "Age", "Fare", "Pclass"]])
```

```python
#2.
print(df.nlargest(10, "Fare")[["Name", "Fare", "Pclass"]])
```

```python
#3
print(df.groupby("Pclass")["Survived"].mean().sort_values(ascending=False))
```

```python
#4.
print(df.groupby("Sex")[["Age", "Fare"]].mean())
```

```python
#5.
print(df[(df["SibSp"] >= 2) & (df["Parch"] >= 1)])
```

```python
#6.
print(df.groupby(["Sex", "Pclass"]).size())
```

```python
#7.
missing = df.isnull().sum()
print(missing[missing > 0])
```

```python
#9.
print(df[df["Fare"] > df["Fare"].mean()][["Name", "Fare", "Pclass"]])
```

```python
#10.
summary = df.groupby("Pclass").agg(
    Total_Passengers=("PassengerId", "count"),
    Survivors=("Survived", "sum"),
)
summary["Survival_Percentage"] = (summary["Survivors"] / summary["Total_Passengers"]) * 100
print(summary)
```
